```python
# snscrape is a social networking service scrapper
import snscrape.modules.twitter as twitterscrapper
import pandas as pd
import datetime as dt
```

```python
date = dt.datetime(2020, 12, 1)
hastag_terms = ['bitcoin OR crypto OR BTC OR cryptocurrency', '#
        blockchain', '#ethereum', '#ETH', '#DeFi'] 
start_date = '2021-01-01'
lang = 'en'
tweets = []
```

```python
# loop to scrape 50000+ Crypto related tweets from january to december 2021
for i in range(365):
    date += dt.timedelta(days=1)
    dates = str(date)
    for i,tweet in enumerate(twitterscrapper.TwitterHashtagScraper(hastag_terms +" lang:"+ lang +" since:"+ start_date +" until:"+ dates).get_items()):
        if i > 200:
            break
        tweets.append([tweet.id, tweet.date, tweet.user.location, tweet.content, tweet.hashtags])
tweets_df = pd.DataFrame(tweets, columns=['Tweet_id', 'Datetime', 'User_location', 'Text', 'Hashtags'])
tweets_df.to_csv('../data/raw/Cryptotweetsdataset.csv', encoding='utf-8-sig', index=False)
```
