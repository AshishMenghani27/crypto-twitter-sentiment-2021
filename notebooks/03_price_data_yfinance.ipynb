# BTC-USD and ETH-USD prices via yfinance

Original file name: `Finance Dataset.ipynb`

**Needs network access** (downloads prices from Yahoo Finance via `yfinance`). The prices it produced are committed in `data/prices/`, so later notebooks do not need this step. Re-running can return slightly different values if Yahoo revises its history.

```python
#libraries
import yfinance as yf
import pandas as pd
import os
```

```python
# Directory to save the CSV files
output_dir = "../data/prices/"
os.makedirs(output_dir, exist_ok=True)

# Parameters
tickers = ['BTC-USD', 'ETH-USD']
intervals = ['1d', '1wk', '1mo']
start_date = '2021-01-01'
end_date = '2021-12-31'
```

```python
# Fetching the data, calculating returns, and saving it as .csv files
saved_files = []
for ticker in tickers:
    for interval in intervals:
        data = yf.Ticker(ticker).history(start=start_date, end=end_date, interval=interval)
        
        # Calculate returns and add it as a new column
        data['Returns'] = (data['Close'] - data['Close'].shift(1)) / data['Close'].shift(1)
        
        file_path = os.path.join(output_dir, f"{ticker}_{interval}_2021.csv")
        data.to_csv(file_path)
        saved_files.append(file_path)

print("Files saved at:", saved_files)
```
